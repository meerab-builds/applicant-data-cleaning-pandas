```python
before_rows = len(df)

df = df.dropna(subset=["Applicant Name", "Email"])

removed_rows = before_rows - len(df)

print("Rows removed:", removed_rows)
print("Remaining rows:", len(df))
```

```python
df["University"] = df["University"].fillna("Not Specified")

df["Status"] = df["Status"].fillna("Under Review")
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

## Removing Duplicates

Exact duplicate rows were removed. Duplicate email addresses were also checked, keeping the first record for each email.

```python
exact_duplicates = df.duplicated().sum()
df = df.drop_duplicates()
email_duplicates = df.duplicated(subset=["Email"]).sum()
df = df.drop_duplicates(subset=["Email"], keep="first")
print("Exact duplicates remaining:", df.duplicated().sum())

print(
    "Duplicate emails remaining:",
    df.duplicated(subset=["Email"]).sum()
)
```

## Standardizing Text

Extra spaces were removed from text fields. Applicant names were formatted consistently, and email addresses were converted to lowercase.

```python
# Remove extra spaces from text columns
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Standardize applicant names
df["Applicant Name"] = df["Applicant Name"].apply(
    lambda x: x.title() if isinstance(x, str) else x
)

# Standardize email addresses
df["Email"] = df["Email"].apply(
    lambda x: x.lower() if isinstance(x, str) else x
)

# Standardize application statuses
df["Status"] = df["Status"].apply(
    lambda x: x.strip().title() if isinstance(x, str) else x
)

print("Text formatting standardized successfully!")
```

```python
# Remove extra spaces from all text columns
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].apply(
        lambda value: value.strip() if isinstance(value, str) else value
    )

# Standardize applicant names
df["Applicant Name"] = df["Applicant Name"].str.title()

# Standardize email addresses
df["Email"] = df["Email"].str.lower()

print("Text standardization completed!")
display(df.head())
```

## Standardizing Application Domains

Different spellings and short forms of application domains were mapped to consistent category names.

```python
# Convert domain values to lowercase and remove extra spaces
df["Domain Applied"] = (
    df["Domain Applied"].str.strip().str.lower()
)

# Create a mapping dictionary
domain_mapping = {
    "web dev": "Web Development",
    "webdev": "Web Development",
    "web development": "Web Development",

    "digital marketing": "Digital Marketing",
    "digital mktg": "Digital Marketing",

    "data sci": "Data Science",
    "data science": "Data Science",

    "data analytics": "Data Analytics",
    "data analysis": "Data Analytics",

    "graphic design": "Graphic Design",
    "graphics": "Graphic Design"
}

# Apply the mapping
df["Domain Applied"] = df["Domain Applied"].map(domain_mapping)

print("Domain standardization completed.")
```

```python
print("Unique domain categories:")
print(sorted(df["Domain Applied"].dropna().unique()))

print("\nNumber of domain categories:")
print(df["Domain Applied"].nunique())

print("\nMissing domain values:")
print(df["Domain Applied"].isnull().sum())
```

## Dates and Phone Numbers

Application dates were converted to datetime format. Phone numbers were treated as text, and spaces and hyphens were removed.

```python
# Convert application dates into datetime format
df["Application Date"] = pd.to_datetime(
    df["Application Date"],
    errors="coerce",
    format="mixed"
)

# Remove dashes and spaces from phone numbers
df["Phone"] = (
    df["Phone"]
    .astype("string")
    .str.replace("-", "", regex=False)
    .str.replace(" ", "", regex=False)
)

print("Updated data types:")
df.info()
```

```python
print("Missing or unreadable dates:")
print(df["Application Date"].isnull().sum())

print("\nPhone number examples:")
print(df["Phone"].head())

print("\nDate examples:")
print(df["Application Date"].head())
```
